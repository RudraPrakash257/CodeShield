```python
"""CodeShield - Student Submission Similarity Analyzer.

A Streamlit application that analyzes student submissions and identifies
potential similarity patterns for human review.
"""
import streamlit as st
import os
from pathlib import Path
from tempfile import NamedTemporaryFile
import numpy as np
from itertools import combinations

# Load environment variables
from dotenv import load_dotenv
load_dotenv()

# Import CodeShield modules
from src.extractors.pdf import extract_text_from_pdf
from src.extractors.docx import extract_text_from_docx
from src.extractors.text import extract_text_from_file
from src.ml.features import extract_pairwise_features
from src.ml.model import load_model
from src.ml.predict import predict_similarity, calculate_heuristic_score
from src.visuals.heatmap import create_similarity_heatmap, create_feature_barchart
from src.visuals.network import create_similarity_network


# Page config
st.set_page_config(
    page_title="CodeShield",
    page_icon="🛡️",
    layout="wide",
    initial_sidebar_state="expanded"
)

# Custom CSS theming
st.markdown("""
<style>
    @import url('https://fonts.googleapis.com/css2?family=Inter:wght@300;400;500;600;700&display=swap');
    html, body, [class*="css"] { font-family: 'Inter', sans-serif; }

    section[data-testid="stSidebar"] {
        background: linear-gradient(180deg, #0f172a 0%, #1e293b 100%);
    }
    section[data-testid="stSidebar"] * { color: #e2e8f0 !important; }
    section[data-testid="stSidebar"] .stMarkdown p { color: #94a3b8 !important; }

    .main .block-container { padding-top: 2rem; max-width: 1200px; }

    .stButton > button[kind="primary"] {
        background: linear-gradient(135deg, #3b82f6, #1d4ed8);
        border: none; color: white; font-weight: 600;
        padding: 0.6rem 1.5rem; border-radius: 8px; transition: all 0.2s;
        box-shadow: 0 4px 14px rgba(59,130,246,0.3);
    }
    .stButton > button[kind="primary"]:hover {
        transform: translateY(-1px);
        box-shadow: 0 6px 20px rgba(59,130,246,0.4);
    }

    [data-testid="metric-container"] {
        background: linear-gradient(135deg, #f8fafc, #f1f5f9);
        border: 1px solid #e2e8f0; border-radius: 12px; padding: 1rem;
        box-shadow: 0 2px 8px rgba(0,0,0,0.06);
    }

    .flagged-badge {
        background: linear-gradient(135deg, #fef2f2, #fee2e2);
        border: 1px solid #fca5a5; color: #dc2626;
        padding: 0.25rem 0.75rem; border-radius: 999px;
        font-size: 0.8rem; font-weight: 600;
    }
</style>
""", unsafe_allow_html=True)


# Constants
SIMILARITY_THRESHOLD = float(os.getenv('SIMILARITY_THRESHOLD', 0.70))


@st.cache_resource
def get_model():
    """Load ML model once."""
    return load_model()


@st.cache_data
def process_uploaded_file(file_bytes: bytes, file_name: str):
    """Process uploaded file and extract text."""
    suffix = Path(file_name).suffix
    with NamedTemporaryFile(delete=False, suffix=suffix) as tmp:
        tmp.write(file_bytes)
        tmp_path = tmp.name

    try:
        ext = suffix.lower()

        if ext == '.pdf':
            return extract_text_from_pdf(tmp_path)

        elif ext == '.docx':
            return extract_text_from_docx(tmp_path)

        elif ext in ['.txt', '.md']:
            return extract_text_from_file(tmp_path)

        else:
            return None

    except Exception as e:
        st.error(f"Error processing {file_name}: {str(e)}")
        return None
    finally:
        try:
            os.unlink(tmp_path)
        except Exception:
            pass


def generate_flag_reasons(features: dict, score: float) -> list:
    """Generate bullet reasons for why a pair was flagged."""
    reasons = []

    if features.get('semantic_similarity', 0) >= 0.7:
        reasons.append("High semantic similarity - shared meaning/topic phrasing")
    elif features.get('semantic_similarity', 0) >= 0.5:
        reasons.append("Moderate semantic similarity detected")

    if features.get('lexical_similarity', 0) >= 0.6:
        reasons.append("Significant word/phrase overlap (TF-IDF cosine)")
    elif features.get('lexical_similarity', 0) >= 0.4:
        reasons.append("Notable lexical alignment")

    if features.get('structural_similarity', 0) >= 0.8:
        reasons.append("Very similar document structure and formatting")
    elif features.get('structural_similarity', 0) >= 0.6:
        reasons.append("Similar document formatting patterns")

    matching_phrases = features.get('matching_phrases', [])
    if len(matching_phrases) >= 3:
        reasons.append(f"{len(matching_phrases)} matching phrases/n-grams found")
    elif len(matching_phrases) >= 1:
        reasons.append("Shared phrases detected")

    if not reasons:
        reasons.append("Combined similarity metrics exceed threshold")

    return reasons[:5]


def generate_summary_paragraph(features: dict, doc_a: dict, doc_b: dict) -> str:
    """Generate human-readable summary paragraph."""
    sem = features.get('semantic_similarity', 0)
    lex = features.get('lexical_similarity', 0)
    struct = features.get('structural_similarity', 0)

    parts = []
    if sem >= 0.7:
        parts.append("strong semantic alignment (high embedding similarity)")
    elif sem >= 0.5:
        parts.append("moderate semantic similarity (embeddings are close)")

    if lex >= 0.6:
        parts.append("significant lexical overlap (many shared words/phrases)")
    elif lex >= 0.4:
        parts.append("some lexical similarity (overlapping vocabulary)")

    if struct >= 0.7:
        parts.append("similar document structure (length, sentence patterns)")

    if parts:
        summary = f"This pair shows {', '.join(parts)}. "
    else:
        summary = "This pair shows some similarity patterns. "

    summary += (
        "These combined signals make it a candidate for manual review. "
        "Note: Similarity alone is not evidence of plagiarism - "
        "shared topics, course materials, templates, or common terminology "
        "can naturally produce these patterns."
    )
    return summary


def render_sidebar(model, metadata):
    """Render sidebar with status indicators."""
    st.sidebar.title("🛡️ CodeShield")
    st.sidebar.caption("Student Submission Similarity Analyzer")
    st.sidebar.markdown("---")
    st.sidebar.subheader("System Status")

    if model:
        st.sidebar.success("✅ ML Model Loaded")
        if metadata:
            st.sidebar.caption(f"Type: {metadata.get('model_type', 'unknown').upper()}")
            acc = metadata.get('metrics', {}).get('accuracy', 0)
            if acc:
                st.sidebar.caption(f"Accuracy: {acc:.1%}")
    else:
        st.sidebar.warning("⚠️ Using Heuristic Scoring")
        st.sidebar.caption("No trained model found")

    st.sidebar.markdown("---")
    st.sidebar.subheader("Settings")
    threshold = st.sidebar.slider(
        "Flag Threshold",
        min_value=0.50, max_value=0.95,
        value=SIMILARITY_THRESHOLD, step=0.05,
        help="Pairs above this score will be flagged for review"
    )
    st.sidebar.markdown("---")
    st.sidebar.caption(
        "Reminder: CodeShield flags similarity patterns for human review. "
        "It does not determine plagiarism."
    )
    return threshold


def main():
    """Main application."""
    model, metadata = get_model()
    threshold = render_sidebar(model, metadata)

    st.title("🛡️ CodeShield — Student Submission Similarity Analyzer")
    st.markdown(
        "Upload student submissions to identify potential similarity patterns. "
        "All findings require human review."
    )

    with st.expander("ℹ️ How to interpret results", expanded=False):
        st.info(
            "CodeShield computes similarity signals — not plagiarism verdicts. "
            "Similarity can occur naturally due to shared topics, common templates, "
            "same reference materials, paraphrasing, expected terminology, and formatting. "
            "Always conduct a manual review before drawing conclusions."
        )

    st.divider()
    st.header("Upload Submissions")

    uploaded_files = st.file_uploader(
        "Upload 2 or more student submissions",
        type=['pdf', 'docx', 'txt', 'md'],
        accept_multiple_files=True,
        help="Supported: PDF, DOCX, TXT, MD"
    )

    if not uploaded_files:
        st.markdown(
            """<div style="text-align:center;padding:3rem;background:#f8fafc;
            border-radius:12px;border:2px dashed #cbd5e1;margin-top:1rem;">
            <h3 style="color:#64748b;">Drop files here to get started</h3>
            <p style="color:#94a3b8;">Supports PDF, DOCX, TXT, MD files</p></div>""",
            unsafe_allow_html=True
        )
        return

    if len(uploaded_files) < 2:
        st.warning("Please upload at least 2 files for comparison.")
        return

    st.success(f"✅ {len(uploaded_files)} files ready for analysis")

    col_btn, col_info = st.columns([1, 3])
    with col_btn:
        analyze_clicked = st.button("🔍 Analyze Submissions", type="primary", use_container_width=True)
    with col_info:
        n_pairs_preview = len(uploaded_files) * (len(uploaded_files) - 1) // 2
        st.caption(f"Will compare {len(uploaded_files)} files → {n_pairs_preview} pairs")

    if not analyze_clicked:
        return

    # Process files
    with st.spinner("Processing submissions..."):
        documents = []
        progress = st.progress(0, text="Extracting text...")
        for i, uf in enumerate(uploaded_files):
            progress.progress((i + 1) / len(uploaded_files), text=f"Processing: {uf.name}")
            doc = process_uploaded_file(uf.getbuffer().tobytes(), uf.name)
            if doc:
                doc['filename'] = uf.name
                documents.append(doc)
        progress.empty()

    if len(documents) < 2:
        st.error("Could not extract text from enough files for comparison.")
        return

    # Compute similarities
    n = len(documents)
    n_pairs = n * (n - 1) // 2
    results = []
    similarity_scores = []
    similarity_matrix = np.zeros((n, n))
    np.fill_diagonal(similarity_matrix, 1.0)

    with st.spinner(f"Computing {n_pairs} pair similarities..."):
        pair_progress = st.progress(0)
        for idx, (i, j) in enumerate(combinations(range(n), 2)):
            doc_a = documents[i]
            doc_b = documents[j]
            features = extract_pairwise_features(doc_a, doc_b)

            if model and metadata:
                score = predict_similarity(model, features, metadata)
                score_type = "ml"
            else:
                score = calculate_heuristic_score(features)
                score_type = "heuristic"

            result = {
                'doc_a': doc_a['filename'],
                'doc_b': doc_b['filename'],
                'index_a': i,
                'index_b': j,
                'score': score,
                'score_type': score_type,
                'features': features,
                'flagged': score >= threshold
            }
            results.append(result)
            similarity_scores.append((i, j, score))
            similarity_matrix[i, j] = score
            similarity_matrix[j, i] = score
            pair_progress.progress((idx + 1) / n_pairs)
        pair_progress.empty()

    results.sort(key=lambda x: x['score'], reverse=True)
    flagged_count = sum(1 for r in results if r['flagged'])
    avg_score = float(np.mean([r['score'] for r in results]))
    max_score = float(np.max([r['score'] for r in results]))

    st.divider()

    # Overview
    st.header("Overview")
    if results and results[0]['score_type'] == 'heuristic':
        st.warning(
            "ML model not found — using heuristic scoring. "
            "Train a model with `python training/train_model.py` for improved accuracy."
        )

    c1, c2, c3, c4, c5 = st.columns(5)
    with c1:
        st.metric("Files Analyzed", len(documents))
    with c2:
        st.metric("Pairs Analyzed", n_pairs)
    with c3:
        st.metric("Flagged Pairs", flagged_count, delta=f">={threshold:.0%}")
    with c4:
        st.metric("Avg Similarity", f"{avg_score:.1%}")
    with c5:
        st.metric("Max Similarity", f"{max_score:.1%}")

    st.divider()

    # Rankings table
    st.header("Top Similar Pairs")
    import pandas as pd

    df_data = []
    for rank, r in enumerate(results[:20], 1):
        df_data.append({
            'Rank': rank,
            'Pair': f"{r['doc_a']} vs {r['doc_b']}",
            'Overall': f"{r['score']:.1%}",
            'Semantic': f"{r['features'].get('semantic_similarity', 0):.1%}",
            'Lexical': f"{r['features'].get('lexical_similarity', 0):.1%}",
            'Structural': f"{r['features'].get('structural_similarity', 0):.1%}",
            'Status': "🚩 Flagged" if r['flagged'] else "✅ OK"
        })

    df = pd.DataFrame(df_data)
    st.dataframe(df, use_container_width=True, hide_index=True)

    st.divider()

    # Visualizations
    viz_col1, viz_col2 = st.columns([3, 2])
    with viz_col1:
        st.subheader("Similarity Heatmap")
        fig_heatmap = create_similarity_heatmap(documents, similarity_matrix)
        st.plotly_chart(fig_heatmap, use_container_width=True)

    with viz_col2:
        st.subheader("Similarity Network")
        st.caption("Nodes = submissions, edges = flagged pairs")
        fig_network = create_similarity_network(documents, similarity_scores, threshold)
        st.plotly_chart(fig_network, use_container_width=True)

    st.divider()

    # Pair Inspector
    st.header("Pair Inspector")
    pair_options = [f"{r['doc_a']} vs {r['doc_b']}" for r in results]
    selected_pair_label = st.selectbox("Select a pair to inspect:", pair_options)

    if not selected_pair_label:
        return

    selected_idx = pair_options.index(selected_pair_label)
    result = results[selected_idx]

    if result['flagged']:
        st.error(f"🚩 FLAGGED — {result['score']:.1%} similarity (above {threshold:.0%} threshold)")
    else:
        st.success(f"✅ Below threshold — {result['score']:.1%} similarity")

    col1, col2 = st.columns(2)
    with col1:
        st.subheader("Similarity Metrics")
        m1, m2, m3 = st.columns(3)
        with m1:
            st.metric("Lexical", f"{result['features'].get('lexical_similarity', 0):.1%}")
        with m2:
            st.metric("Semantic", f"{result['features'].get('semantic_similarity', 0):.1%}")
        with m3:
            st.metric("Structural", f"{result['features'].get('structural_similarity', 0):.1%}")

        st.metric("Overall Score", f"{result['score']:.1%}")

        if result['score_type'] == 'ml':
            st.caption("Score from trained ML model")
        else:
            st.caption("Heuristic: 0.30×lexical + 0.50×semantic + 0.20×structural")

    with col2:
        st.subheader("Feature Breakdown")
        fig_features = create_feature_barchart(result['features'])
        st.plotly_chart(fig_features, use_container_width=True)

    if result['flagged']:
        st.subheader("Why Flagged?")
        fc1, fc2 = st.columns(2)
        with fc1:
            st.markdown("**Key signals:**")
            reasons = generate_flag_reasons(result['features'], result['score'])
            for reason in reasons:
                st.markdown(f"- {reason}")
        with fc2:
            matching = result['features'].get('matching_phrases', [])
            if matching:
                st.markdown("**Matching phrases:**")
                for phrase in matching[:5]:
                    st.markdown(f'> "{phrase}"')

        st.subheader("Human-Review Summary")
        summary = generate_summary_paragraph(
            result['features'],
            documents[result['index_a']],
            documents[result['index_b']]
        )
        st.info(summary)

    st.divider()
    st.subheader("Document Previews")
    pc1, pc2 = st.columns(2)
    with pc1:
        st.markdown(f"**{result['doc_a']}**")
        text_a = documents[result['index_a']].get('text', '')
        preview_a = (text_a[:1500] + "\n[... truncated ...]") if len(text_a) > 1500 else text_a
        st.text_area("Content A", preview_a, height=250, key="preview_a", label_visibility="collapsed")
    with pc2:
        st.markdown(f"**{result['doc_b']}**")
        text_b = documents[result['index_b']].get('text', '')
        preview_b = (text_b[:1500] + "\n[... truncated ...]") if len(text_b) > 1500 else text_b
        st.text_area("Content B", preview_b, height=250, key="preview_b", label_visibility="collapsed")

    st.divider()
    st.subheader("Export Results")
    export_data = []
    for r in results:
        export_data.append({
            'Pair_A': r['doc_a'],
            'Pair_B': r['doc_b'],
            'Overall_Score': round(r['score'], 4),
            'Semantic_Similarity': round(r['features'].get('semantic_similarity', 0), 4),
            'Lexical_Similarity': round(r['features'].get('lexical_similarity', 0), 4),
            'Structural_Similarity': round(r['features'].get('structural_similarity', 0), 4),
            'Flagged': r['flagged'],
            'Scoring_Method': r['score_type']
        })
    export_df = pd.DataFrame(export_data)
    csv_data = export_df.to_csv(index=False)
    st.download_button(
        label="Download Results as CSV",
        data=csv_data,
        file_name="codeshield_results.csv",
        mime="text/csv"
    )


if __name__ == "__main__":
    main()
```
